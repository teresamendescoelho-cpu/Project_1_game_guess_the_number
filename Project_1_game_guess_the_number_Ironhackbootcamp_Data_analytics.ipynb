```python
import random


def choose_difficulty():
    """
    Lets the player choose the difficulty level.

    Classic follows the original Ironhack project requirement:
    numbers from 1 to 100 with a maximum of 10 attempts.

    Easy uses numbers from 1 to 50 with 10 attempts.

    Hard uses numbers from 1 to 100 with 8 attempts.

    Returns:
        tuple: maximum number, maximum attempts, and difficulty name.
    """

    while True:
        print("\nChoose your difficulty:")
        print("1 - Easy    (1-50, 10 attempts)")
        print("2 - Classic (1-100, 10 attempts)")
        print("3 - Hard    (1-100, 8 attempts)")

        choice = input("Enter your choice (1, 2, or 3): ").strip()

        if choice == "1":
            return 50, 10, "Easy"

        elif choice == "2":
            return 100, 10, "Classic"

        elif choice == "3":
            return 100, 8, "Hard"

        else:
            print("Please choose 1, 2, or 3.")


def get_player_guess(min_number, max_number):
    """
    Gets and validates the player's guess.

    Invalid text or numbers outside the allowed range
    do not count as an attempt.

    Returns:
        int: A valid guess.
    """

    while True:
        try:
            guess = int(
                input(
                    f"Enter your guess ({min_number}-{max_number}): "
                )
            )

            if guess < min_number or guess > max_number:
                print(
                    f"Please enter a number between "
                    f"{min_number} and {max_number}."
                )
                continue

            return guess

        except ValueError:
            print("Please enter a valid number.")


def check_guess(guess, secret_number):
    """
    Compares the player's guess with the secret number.

    Returns:
        str: Feedback indicating whether the guess is too high,
        too low, or correct.
    """

    if guess < secret_number:
        return "Too low!"

    elif guess > secret_number:
        return "Too high!"

    else:
        return "Correct!"


def update_game_state(game_state, guess):
    """
    Updates the game state after a valid guess.

    The guess is added to the list and the number of
    attempts used is increased by one.
    """

    game_state["guesses"].append(guess)
    game_state["attempts_used"] += 1
    game_state["remaining_attempts"] -= 1


def display_game_status(game_state):
    """
    Displays the current game progress.
    """

    print(
        f"Attempts used: {game_state['attempts_used']} | "
        f"Attempts remaining: {game_state['remaining_attempts']}"
    )

    print(f"Previous guesses: {game_state['guesses']}")


def check_game_over(game_state):
    """
    Checks whether the player has won or lost.

    Returns:
        bool: True if the game is over, otherwise False.
    """

    if game_state["won"]:
        return True

    if game_state["remaining_attempts"] == 0:
        return True

    return False


def play_one_game():
    """
    Runs one complete game.

    The player chooses a difficulty and tries to guess
    the randomly generated secret number.

    A for loop controls the maximum number of attempts.

    Returns:
        bool: True if the player won, False if the player lost.
    """

    print("\n" + "=" * 45)
    print("      NUMBER GUESSING GAME")
    print("=" * 45)

    max_number, max_attempts, difficulty = choose_difficulty()

    secret_number = random.randint(1, max_number)

    game_state = {
        "secret_number": secret_number,
        "difficulty": difficulty,
        "attempts_used": 0,
        "remaining_attempts": max_attempts,
        "guesses": [],
        "won": False
    }

    print("\n" + "-" * 45)
    print(f"Difficulty: {difficulty}")
    print(f"I'm thinking of a number between 1 and {max_number}.")
    print(f"You have {max_attempts} attempts.")
    print("-" * 45)

    # The for loop controls the maximum number of attempts.
    for attempt in range(max_attempts):

        if check_game_over(game_state):
            break

        guess = get_player_guess(1, max_number)

        update_game_state(game_state, guess)

        feedback = check_guess(
            guess,
            game_state["secret_number"]
        )

        print(f"\n{feedback}")

        # Display the status after every valid guess.
        display_game_status(game_state)

        if feedback == "Correct!":
            game_state["won"] = True

            print(
                f"\nYou win! "
                f"You guessed the number in "
                f"{game_state['attempts_used']} attempts."
            )

            break

    if not game_state["won"]:
        print("\nGame over!")
        print(
            f"The number was "
            f"{game_state['secret_number']}."
        )

    print("\nFinal game statistics:")
    print(f"Difficulty: {game_state['difficulty']}")
    print(f"Attempts used: {game_state['attempts_used']}")
    print(f"Your guesses: {game_state['guesses']}")

    return game_state["won"]


def play_again():
    """
    Asks the player whether they want to play another game.

    Returns:
        bool: True if the player wants to continue,
        False if they want to stop.
    """

    while True:
        choice = input(
            "\nWould you like to play again? (y/n): "
        ).strip().lower()

        if choice == "y":
            return True

        elif choice == "n":
            return False

        else:
            print("Please enter 'y' for yes or 'n' for no.")


def number_guessing_game():
    """
    Runs a complete and playable Number Guessing Game.

    The player can choose a difficulty level and tries to guess
    a randomly generated number within the selected range.

    The Classic difficulty follows the original Ironhack project
    requirement: guess a number between 1 and 100 with a maximum
    of 10 attempts.

    The Easy difficulty uses numbers from 1 to 50 with 10 attempts.

    The Hard difficulty uses numbers from 1 to 100 with 8 attempts.

    The program uses:
    - A list to store the player's guesses.
    - A dictionary to track the game state.
    - Functions to organize the game into reusable components.
    - A for loop to control the maximum number of attempts.
    - While loops for input validation and repeated gameplay.
    - Conditional statements for decision-making.
    - The random module to generate the secret number.
    - Try/except for input validation and error handling.
    - Docstrings to document the main functions.

    The player can also choose to play again after each completed
    game. At the end of the session, the program displays the
    number of games played, games won, and the overall win rate.
    """

    print("\nWelcome to the Number Guessing Game!")

    games_played = 0
    games_won = 0

    while True:

        won = play_one_game()

        games_played += 1

        if won:
            games_won += 1

        if not play_again():
            break

    print("\n" + "=" * 45)
    print("          THANK YOU FOR PLAYING!")
    print("=" * 45)

    print(f"Games played: {games_played}")
    print(f"Games won: {games_won}")

    if games_played > 0:
        win_rate = (games_won / games_played) * 100
        print(f"Win rate: {win_rate:.1f}%")

    print("Goodbye!")


number_guessing_game()
```
