```python
import random


def number_guessing_game():
    """
    Runs a playable number guessing game.

    The player has 10 attempts to guess a random number between 1 and 100.
    The program uses a list to store guesses, a dictionary to store game
    statistics, loops for repeated attempts, flow control for decisions,
    the random module, and try/except for error handling.
    """

    print("Welcome to the Number Guessing Game!")
    print("I'm thinking of a number between 1 and 100.")
    print("You have 10 attempts to guess it.")

    secret_number = random.randint(1, 100)
    max_attempts = 10

    guesses = []

    game_stats = {
        "attempts_used": 0,
        "won": False
    }

    for attempt in range(1, max_attempts + 1):
        try:
            guess = int(input("Enter your guess: "))
        except ValueError:
            print("Please enter a valid number.")
            continue

        guesses.append(guess)
        game_stats["attempts_used"] += 1

        if guess < 1 or guess > 100:
            print("Please guess a number between 1 and 100.")
            continue

        if guess < secret_number:
            print("Too low!")
        elif guess > secret_number:
            print("Too high!")
        else:
            game_stats["won"] = True
            print(f"You win! You guessed the number in {game_stats['attempts_used']} attempts.")
            break

        print(f"Attempts left: {max_attempts - game_stats['attempts_used']}")

    if not game_stats["won"]:
        print(f"Out of attempts! The number was {secret_number}.")
        print("Better luck next time!")

    print(f"Your guesses were: {guesses}")
    print(f"Game statistics: {game_stats}")


number_guessing_game()
```
